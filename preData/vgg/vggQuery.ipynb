```python
import numpy as np
import h5py
import matplotlib.pyplot as plt
from VGG_feature_extractor import VGGNet
from scipy import spatial
```

```python
h5f = h5py.File("VGG16Features.h5", 'r')
feats = h5f['dataset_1'][:]
imgNames = h5f['dataset_2'][:]
h5f.close()
```

```python
queryImg = "query_images/monkey4.jpg"
```

```python
model = VGGNet()
X = model.extract_feat(queryImg)
```

```python
scores = []
for i in range(feats.shape[0]):
    score = 1 - spatial.distance.cosine(X, feats[i])
    scores.append(score)
scores = np.array(scores)
```

```python
rank_ID = np.argsort(scores)[::-1]
rank_score = scores[rank_ID]

maxres = 10
imlist = [imgNames[index] for index in rank_ID[0:maxres]]

for i, img_name in enumerate(imlist):
    img = plt.imread(f"image_directory/{img_name}")
    plt.subplot(1, maxres, i+1)
    plt.imshow(img)
    plt.title(f"Match {i+1}")
    plt.axis('off')
plt.show()
```
