```python
import joblib
import os
import tensorflow as tf
```

```python
def convert_sav_to_h5(sav_file_path):
    model = joblib.load(sav_file_path)
    if isinstance(model, tf.keras.Model):
        h5_file_path = sav_file_path.replace('.sav', '.h5')
        model.save(h5_file_path)

sav_file_path = 'vggFinetune.sav'
convert_sav_to_h5(sav_file_path)
```
